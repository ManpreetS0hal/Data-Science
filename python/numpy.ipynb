```python
def myFunc(a,b):
    "Return a-b if a>b,otherwise return a+b"
    if a> b:
        return a -b 
    else:
        return a + b
```

```python
vFunc = np.vectorize(myFunc)
vFunc([1,2,3,4,5],2)
```

```python
# STACKING

A = np.arange(1,5).reshape(2,2)
B = np.arange(11,15).reshape(2,2)
print(A)
print(B)
```
