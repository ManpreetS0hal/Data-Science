```python
#Write a script that takes a user’s birth year and calculates their age, handling potential non-integer inputs.
def age(year):
    if(year < 2026 ):
        present_age = 2026 - year
    else:
        present_age = "Not born yet"
    return present_age

age(2003)
age(2027)
```

```python
# Python3 program to display Prime numbers till 100

def isPrime(n):
    if(n==1 or n==0):  return False
    for i in range(2,n):
        if(n%i==0):
            return False

    return True

N = 100;
for i in range(1,N+1):
    if(isPrime(i)):
        print(i,end=" ")
```

```python
# Fibonacci series

def fibb(n):
    if(n <= 1):
        return n
    else:
        return fibb(n - 1) + fibb(n - 2)

print(fibb(15))
```

```python
#Given a list of integers, write a function to return the second largest element without using .sort().

ls=[5,2,4,8,6,9]

largest = float('-inf')
second= float('-inf')

for num in ls:
    if num > largest:
        second = largest
        largest = num
    elif num > second and num != largest :
        second = num

print(second)
```

```python
#Write a script to remove all duplicate elements from a list while maintaining the original order.

ls = [3,3,5,5,4,4,4,4,7,7]
ls2 =[]

for num in ls:
    if num not in ls2:
        ls2.append(num)
        
print(ls2)
```

```python
#Create a dictionary representing a store inventory. Write a function to update prices by a percentage and another to calculate the total stock quantity.

store_invent = {
    'apple' : {'price' : 110, 'quantity': 3000},
    'strawberry' : {'price' : 220, 'quantity': 2500},
    'guava' : {'price' : 90, 'quantity':1000}
}

def update_prices(store_invent,percentage1):
    for item in store_invent:
        store_invent[item]['price'] += store_invent[item]['price'] * (percentage1 / 100)

def total_quantity(store_invent):
    total = 0
    for item in store_invent:
        total += store_invent[item]['price'] * store_invent[item]['quantity']
    return total

update_prices(store_invent, 90)

print(store_invent)

print("The total stock value is: ", total_quantity(store_invent))
```
